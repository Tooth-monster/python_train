欧几里得算法

```python
def gcd(m,n):
    while n != 0:
        m, n = n, m % n
    return m
print(gcd(2018, 2019))   
print(gcd(1200, 300))    
print(gcd(5040, 60))     

print(gcd(-12, -8))
print(gcd(-12,8))

```

如果⼀个或两个参数是负数，你的函数会做什么？这种⾏为有意义吗？
答：函数返回正数 / 负数取决于 m 符号；数学上 GCD 定义永远非负。> 欧几里得算法直接传入负数行为不一定符合数学定义，因为模运算对负数在 Python 中会保留符号相关结果，数学 GCD 应该取绝对值。

近似欧拉数e

```python
import math
def euler_limit(n):
    return (1 + 1 / n) ** n

# 测试
print(euler_limit(10000))

def euler_infinite_sum(n):
    if n == 0:
        return 0.0
    total = 0.0
    fact = 1  # 阶乘，初始0! = 1
    for k in range(n):
        if k > 0:
            fact = fact * k
        total += 1.0 / fact
    return total

#测试
print(euler_infinite_sum(1)) #1.0
print(euler_infinite_sum(4)) #≈2.666666

def euler_approx(epsilon):
    e_true = math.exp(1)
    total = 0.0
    fact = 1
    k = 0
    while abs(total - e_true) >= epsilon:
        if k>0:
            fact *= k
        total += 1.0/fact
        k +=1
    return total

#测试
print(euler_approx(0.001))
print(math.exp(1))

def print_euler_sum_table(n):
    for k in range(1, n+1):
        val = euler_infinite_sum(k)
        print(f"k={k}, sum={val}")

#测试
print_euler_sum_table(6)
```

测试整数属性

```python
def is_power_of_2(n):
    while n > 1:
        if n % 2 != 0:
            return False
        n = n // 2
    return True

#测试
print(is_power_of_2(1))   
print(is_power_of_2(8))   
print(is_power_of_2(12))  

def is_power(b, n):
    # n = b^p
    while n % b == 0:
        n = n // b
    #循环结束后，如果n==1说明可以完全被b除尽
    return n == 1

#测试
print(is_power(3, 9))  
print(is_power(2,16))
print(is_power(5,7)) 

```

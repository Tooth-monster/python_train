第一个代码框（身份与环境记录）

```python
import os, uuid, hashlib, platform, getpass

# ===== 1. 填写学号（务必改成自己的）=====
STUDENT_ID = 20244080203        # ← 改这里

# ===== 2. 打印 conda 环境 =====
CONDA_ENV = os.environ.get("CONDA_DEFAULT_ENV", "base")
print(f"[Conda 环境] {CONDA_ENV}")

# ===== 3. 机器指纹（绑定当前电脑）=====
MACHINE_FP = hashlib.sha256(
    f"{platform.node()}|{platform.machine()}|{platform.system()}|{uuid.getnode()}".encode()
).hexdigest()[:16]

# ===== 4. 生成唯一 ID =====
RUN_ID = f"{STUDENT_ID}-{MACHINE_FP}"
print(f"[学号]     {STUDENT_ID}")
print(f"[机器指纹] {MACHINE_FP}")
print(f"[RUN_ID]   {RUN_ID}")
```

实验一：基本运算、数据类型与字符串

```python
#分别计算 8/3、8//3、8%3、2**7，输出结果及其类型，说明运算符含义。
print(8 / 3,  type(8 / 3))    # 2.6666666666666665   真除法（结果总是 float）
print(8 // 3, type(8 // 3))   # 2                      整除（向下取整）
print(8 % 3,  type(8 % 3))    # 2                      取模（余数）
print(2 ** 7, type(2 ** 7))   # 128                    幂运算
```

```python
#创建整数、浮点数、字符串、布尔值各一个，用 type() 检查类型。
a=6
b=3.14
c='hello'
d=True
print(type(a))
print(type(b))  
print(type(c))
print(type(d))
```

```python
#执行 int("123")、float("3.14")、int(3.9)，解释结果；捕获 int("goat") 的异常并说明原因。
print(int("123"))    # 123  字符串数字 → 整数
print(float("3.14")) # 3.14 字符串 → 浮点
print(int(3.9))      # 3    浮点 → 整数是“截断”小数部分（非四舍五入）

try:
    int("goat")
except ValueError as e:
    print("ValueError:", e)   # invalid literal for int() with base 10: 'goat'
# 原因：字符串内容不是合法的十进制整数，无法转换
```

```python
#对 " Python Data Analysis " 做去首尾空白、大小写转换、切片、单词拆分，再用 join() 连接。
t = "  Python Data Analysis  "
st = t.strip()                      # 去掉首尾空白
print(st.upper())                   # PYTHON DATA ANALYSIS
print(st.lower())                   # python data analysis
print(st.title())                   # Python Data Analysis
print(st[0:6])                      # Python   （切片：前 6 个字符）
print(st[-8:])                      # Analysis （切片：后 8 个字符）
words = st.split()                  # 按空白拆分成列表
print(words)                        # ['Python','Data','Analysis']
print("_".join(words))              # Python_Data_Analysis
```

```python
#比较字符串拼接与重复；验证字符串不可直接改单个字符，并给出构造新字符串的方式。
print("Py" + "thon")   # 拼接 -> Python
print("ab" * 3)        # 重复 -> ababab

s = "Python"
try:
    s[0] = "J"         # 字符串不可变
except TypeError as e:
    print("TypeError:", e)  

new_s = "J" + s[1:]    # 切片后重新拼接，生成新字符串
print(new_s)          
```

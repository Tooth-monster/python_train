实验二:函数、条件、循环与递归

```python
#2-1　编写 score_level(score)：90~100 优秀，80~不足90 良好，60~不足80 及格，0~不足60 不及格；超范围抛 ValueError。
def score_level(score):
    if not isinstance(score, (int, float)) or isinstance(score, bool):
        raise ValueError("分数必须是数值")
    if score < 0 or score > 100:
        raise ValueError(f"分数超出 0~100 范围: {score}")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

for s in (95, 85, 70, 59):
    print(s, score_level(s))

try:
    score_level(120)
except ValueError as e:
    print("ValueError:", e)


```

```python
#2-2　用 for 和 while 分别计算 1 到 100 的整数和，比较结果。
total_for = 0
for i in range(1, 101):
    total_for += i

total_while, i = 0, 1
while i <= 100:
    total_while += i
    i += 1

print(total_for, total_while, total_for == total_while)  
```

```python
#2-3　编写递归版与迭代版阶乘（0! = 1，输入非负整数），比较 n = 0、1、5、10 的结果。 
def fact_iter(n):
    if n < 0:
        raise ValueError("n 必须为非负整数")
    r = 1
    for k in range(2, n + 1):
        r *= k
    return r

def fact_rec(n):
    if n < 0:
        raise ValueError("n 必须为非负整数")
    if n == 0:
        return 1
    return n * fact_rec(n - 1)

for n in (0, 1, 5, 10):
    print(n, fact_iter(n), fact_rec(n), fact_iter(n) == fact_rec(n))
```

```python
#2-4　编写 summary(values)，以元组返回最小值、最大值、平均值，并用元组解包接收。
def summary(values):
    if not values:
        raise ValueError("列表不能为空")
    return min(values), max(values), sum(values) / len(values)

mn, mx, avg = summary([85, 58, 92, 76])
print(mn, mx, round(avg, 2))   
```

```python
#2-5　编写 sum_all(*args)，测试无参数、单参数、多参数三种情况。
def sum_all(*args):
    return sum(args)

print(sum_all())            # 0
print(sum_all(5))           # 5
print(sum_all(1, 2, 3, 4))  # 10
```

实验三：列表、字典、元组与基础统计

```python
names  = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]
#3-1　用 zip() 构建姓名→成绩字典，用 items() 逐项输出。
score_map = dict(zip(names, scores))
for name, sc in score_map.items():
    print(name, sc)

print("\n")
#3-2　用 enumerate() 输出序号（从 1 开始）、姓名、成绩。
for idx, (name, sc) in enumerate(zip(names, scores), start=1):
    print(idx, name, sc)

print("\n")
#3-3　用列表推导式提取及格学生姓名，并生成所有成绩的平方列表。
passed    = [n for n, s in zip(names, scores) if s >= 60]
squares   = [s ** 2 for s in scores]
print(passed)   
print(squares)

print("\n")
#3-4　计算人数、平均分、最高分、最低分、及格率；平均分保留两位小数，及格率显示为百分数。
n        = len(scores)
avg      = sum(scores) / n
highest  = max(scores)
lowest   = min(scores)
pass_rate = sum(1 for s in scores if s >= 60) / n

print(f"人数：{n}，平均分：{avg:.2f}，最高分：{highest}，最低分：{lowest}，及格率：{pass_rate:.1%}")

print("\n")
#3-5　比较 alias = scores 与 copied = scores.copy()：改 alias 的一个元素，观察变化。用单独的演示列表，别动原始 scores。
demo   = [1, 2, 3, 4]     # 单独演示列表
alias  = demo             # 引用同一对象（别名）
copied = demo.copy()      # 浅拷贝，新对象

alias[0] = 100

print("demo  :", demo)   
print("alias :", alias)   
print("copied:", copied)  
# 结论：alias 只是同一个列表的另一个名字；copy() 才是独立副本

print("\n")
#3-6　用元组保存一条学生记录，验证元素不能重新赋值；观察长度不同的两个序列 zip() 的结果。
record = ("Alice", 85, "女")

try:
    record[0] = "Bob"          # 元组不可变
except TypeError as e:
    print("TypeError:", e)     

print(list(zip([1, 2, 3, 4], "abc")))
```

实验四：文件读写、异常处理与模块

```python
#4-1 用 os.getcwd() 看工作目录，用 os.path.join() 构造路径，用 os.makedirs(..., exist_ok=True) 建目录。
import os
print(os.getcwd())

data_dir = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(data_dir, exist_ok=True)
print(data_dir)
print("\n")


#4-2 用 with open(..., encoding="utf-8") 写入数据，再读取并跳过表头、逐条解析。
DATA = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""
txt_path = os.path.join(data_dir, "students.txt")

# 写入（w 模式会覆盖原文件）
with open(txt_path, "w", encoding="utf-8") as f:
    f.write(DATA)

# 读取
with open(txt_path, "r", encoding="utf-8") as f:
    lines = f.read().splitlines()

print(lines[0])          # 表头
print(lines[1:])         # 各条记录
print("\n")

#4-3　编写 parse_student(line)：转整数、检查字段数、姓名是否为空、成绩是否 0~100；无效记录抛 ValueError。
def parse_student(line):
    parts = line.strip().split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数量错误（应为 2 个，实际 {len(parts)} 个）")

    name    = parts[0].strip()
    score_s = parts[1].strip()

    if name == "":
        raise ValueError("姓名不能为空")

    try:
        score = int(score_s)
    except ValueError:
        raise ValueError(f"成绩不是合法整数：{score_s!r}")

    if score < 0 or score > 100:
        raise ValueError(f"成绩超出 0~100 范围：{score}")

    return {"name": name, "score": score}
print("\n")

#4-4　逐行处理，捕获异常并记录行号与原因，继续处理后续；有效记录存为字典列表。

valid, errors = [], []

with open(txt_path, "r", encoding="utf-8") as f:
    next(f)                                  # 跳过表头
    for lineno, raw in enumerate(f, start=2):  # 行号从 2 开始（表头是第 1 行）
        line = raw.strip()
        if line == "":
            continue
        try:
            valid.append(parse_student(line))
        except ValueError as e:
            errors.append((lineno, line, str(e)))

print("有效记录：", valid)
print("无效记录：")
for ln, text, msg in errors:
    print(f"  第{ln}行 [{text}] -> {msg}")
print("\n")

#4-5　把有效数、无效数、平均分、及格率写入 exp01_data/report.txt；分别演示 w 与 a 模式的区别。
n_valid  = len(valid)
n_invalid = len(errors)
avg = sum(r["score"] for r in valid) / n_valid
rate = sum(1 for r in valid if r["score"] >= 60) / n_valid

report_path = os.path.join(data_dir, "report.txt")

# w 模式：覆盖写入，已有内容全部清空
with open(report_path, "w", encoding="utf-8") as f:
    f.write(f"有效记录数：{n_valid}\n")
    f.write(f"无效记录数：{n_invalid}\n")
    f.write(f"平均分：{avg:.2f}\n")
    f.write(f"及格率：{rate:.1%}\n")

# a 模式：追加写入，保留原有内容
with open(report_path, "a", encoding="utf-8") as f:
    f.write("=== 追加说明（a 模式）===\n")

print(open(report_path, encoding="utf-8").read())
print("\n")

#4-6　用 pickle 保存有效记录为 exp01_data/students.pkl，再读回验证与原数据相等。
import pickle
pkl_path = os.path.join(data_dir, "students.pkl")
with open(pkl_path, "wb") as f:      # 二进制写
    pickle.dump(valid, f)
with open(pkl_path, "rb") as f:      # 二进制读
    loaded = pickle.load(f)
print(loaded == valid)        
print("\n")

#4-7　把 parse_student() 存到 Notebook 同级的 student_utils.py，用 import 调用，各展示一个有效/无效测试。
# 生成模块文件 student_utils.py
module_src = '''def parse_student(line):
    parts = line.strip().split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数量错误（应为 2 个，实际 {len(parts)} 个）")
    name, score_s = parts[0].strip(), parts[1].strip()
    if name == "":
        raise ValueError("姓名不能为空")
    try:
        score = int(score_s)
    except ValueError:
        raise ValueError(f"成绩不是合法整数：{score_s!r}")
    if score < 0 or score > 100:
        raise ValueError(f"成绩超出 0~100 范围：{score}")
    return {"name": name, "score": score}
'''

with open(os.path.join(os.getcwd(), "student_utils.py"), "w", encoding="utf-8") as f:
    f.write(module_src)

# 导入并测试
from student_utils import parse_student

print(parse_student("Alice,85"))      # 有效 -> {'name':'Alice','score':85}

try:
    parse_student("Henry,105")        # 无效 -> 抛 ValueError
except ValueError as e:
    print("ValueError:", e)
print("\n")

#4-8　读取一个明确不存在的文件，捕获 FileNotFoundError 并输出说明。
missing = os.path.join(data_dir, "not_exist.txt")
try:
    with open(missing, "r", encoding="utf-8") as f:
        f.read()
except FileNotFoundError as e:
    print("文件不存在：", e)
```

实验五：类、继承与对象复制

```python
#5-1　定义 Student 类：初始化 name/score，实现 is_passed() 与 __str__()。
class Student:
    def __init__(self, name, score):
        self.name  = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        status = "及格" if self.is_passed() else "不及格"
        return f"Student(name={self.name}, score={self.score}, {status})"

students = [Student(r["name"], r["score"]) for r in valid]
#5-2　为实验 4 中每条有效记录创建 Student 对象，输出信息与及格状态。
for stu in students:
    print(stu, "| is_passed:", stu.is_passed())
print("\n")

#5-3　定义继承自 Student 的 GraduateStudent：增加 research_topic，用 super().__init__()，重写 __str__()。
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)      # 初始化父类属性
        self.research_topic = research_topic

    def __str__(self):
        status = "及格" if self.is_passed() else "不及格"
        return (f"GraduateStudent(name={self.name}, score={self.score}, "
                f"topic={self.research_topic}, {status})")

gs = GraduateStudent("Alice", 85, "机器学习")
print(gs)
print(isinstance(gs, Student))   # True（继承关系）

#5-4　对嵌套对象 {"name":"Alice","scores":[85,90]} 分别做直接赋值、copy.copy()、copy.deepcopy()，修改内部列表并比较。每次比较都从新建的原始对象开始。
import copy

# ---- 场景 A：直接赋值 ----
orig = {"name": "Alice", "scores": [85, 90]}
a = orig
a["scores"].append(100)
print("A 直接赋值 ->", "orig:", orig["scores"], "a:", a["scores"])
# 两者都是 [85,90,100]：指向同一对象

# ---- 场景 B：浅拷贝 ----
orig = {"name": "Alice", "scores": [85, 90]}      # 重新新建
b = copy.copy(orig)
b["scores"].append(100)
print("B 浅拷贝   ->", "orig:", orig["scores"], "b:", b["scores"])
# orig 也变成 [85,90,100]：外层是新字典，但内层列表仍共享

# ---- 场景 C：深拷贝 ----
orig = {"name": "Alice", "scores": [85, 90]}      # 重新新建
c = copy.deepcopy(orig)
c["scores"].append(100)
print("C 深拷贝   ->", "orig:", orig["scores"], "c:", c["scores"])
# orig 保持 [85,90]，c 为 [85,90,100]：内外层全部独立

```

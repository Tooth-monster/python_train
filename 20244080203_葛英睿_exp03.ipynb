```python
#实验一
import re

text1 = "cat on mat; later: cats and dogs"
text2 = "cat dog bird CAT"
text3 = "ct cat caat caaat"
print("\n实验1.1")
print("待匹配文本已就绪")
print("数字      :", re.findall(r"\d+", text1))
print("空白      :", re.findall(r"\s+", text1))
print("独立词cat :", re.findall(r"\bcat\b", text1))

# 对照：只要求左边界 / 只要求右边界，理解 \b 的作用
print("左边界cat :", re.findall(r"\bcat", text1))
print("右边界cat :", re.findall(r"cat\b", text1))

print("\n实验1.2")
print("cat|dog          :", re.findall(r"cat|dog", text2))
print("忽略大小写        :", re.findall(r"cat|dog", text2, re.I))

print("a|aa|aaa  on 'aaaa':", re.findall(r"a|aa|aaa", "aaaa"))
print("aaa|aa|a  on 'aaaa':", re.findall(r"aaa|aa|a", "aaaa"))
print("\n实验1.3")
pats = [r"ca*t", r"ca+t", r"ca?t", r"ca{2}t", r"ca{1,2}t"]
for p in pats:
    print(f"{p:<10} -> {re.findall(p, text3)}")
```

```python
#实验二
print("实验2.1")
mail = "联系：alice@example.com；备选：bob@test.org"
pat  = r"([\w.+-]+)@([\w-]+\.[\w.-]+)"

print("findall :", re.findall(pat, mail))
for m in re.finditer(pat, mail):
    print(f"整体={m.group(0):<18} 用户名={m.group(1):<6} 域名={m.group(2)}")

# 命名分组写法，结果一致，可读性更好
print("命名分组:", re.findall(r"(?P<user>[\w.+-]+)@(?P<domain>[\w-]+\.[\w.-]+)", mail))
print("\n实验2.2")
dup = "cat cat, dog cat, bird bird"
pat = r"\b(\w+)\s+\1\b"

print("findall :", re.findall(pat, dup))
for m in re.finditer(pat, dup):
    print(f"匹配 {m.group(0)!r:<12} 位置 {m.span()}  捕获组=\\1 → {m.group(1)}")

print("\n实验2.3")
dates = "2024-05-01\n2024/05/02\n2024.05.03"
pat   = r"(\d{4})([-/.])(\d{2})\2(\d{2})"

new = re.sub(pat, r"\4-\3-\1", dates)

print("—— 转换前 ——")
print(dates)
print("—— 转换后 ——")
print(new)
print("行数：", dates.count("\n") + 1, "->", new.count("\n") + 1)

print("\n实验2.4")
tel = "联系我：123-456-7890 或 987.654.3210"

# 写法 A：只接受 - 和 . 两种分隔符，最严谨
print(re.sub(r"\b(\d{3})[-.](\d{3})[-.](\d{4})\b", r"(\1) \2-\3", tel))

# 写法 B：\D 表示任意非数字，更宽容
print(re.sub(r"\b(\d{3})\D(\d{3})\D(\d{4})\b", r"(\1) \2-\3", tel))
```

```python
#实验三
from bs4 import BeautifulSoup

html_doc = """
<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲：正则表达式</p>
<p class="course">第七讲：结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>
"""

soup = BeautifulSoup(html_doc, "html.parser")
print(type(soup))          # <class 'bs4.BeautifulSoup'>

print("\n实验3.1")
print("title 文本 :", soup.title.text)
print("h1 标签名  :", soup.h1.name)
print("h1 文本    :", soup.h1.text)
print("节点类型   :", type(soup.title))

print("\n实验3.2")
ps = soup.find_all("p", class_="course")
print("course 段落数:", len(ps))
for i, p in enumerate(ps, 1):
    print(f"[{i}] class={p.get('class')}  文本={p.get_text()}")

links = soup.find_all("a")
print("链接数:", len(links))
for i, a in enumerate(links, 1):
    print(f"[{i}] 文字={a.text}  href={a.get('href')}")

print("\n实验3.3")
print("整个文档 原始      :", repr(soup.get_text()))
print("整个文档 去首尾空白:", repr(soup.get_text().strip()))
print()
print("整个文档 整理输出:")
print(soup.get_text("\n", strip=True))
print()
print("单个段落:", repr(ps[0].get_text()))
print("单个链接:", repr(links[0].get_text()))
print("仅 body :", repr(soup.body.get_text("\n", strip=True)))
```

### 实验 1 匹配规则说明

- `\d+`：一个或多个数字。原句中不含数字，故返回空列表，属正确结果。
- `\s+`：一个或多个空白字符，7 个单词之间共 6 处空格，故得到 6 个匹配。
- `\bcat\b`：`\b` 为零宽单词边界，不消耗字符，因此 `cats` 不被命中，只有独立的 `cat` 符合。
- `cat|dog`：交替默认区分大小写，`CAT` 需配合 `re.I` 或写成 `[Cc][Aa][Tt]`。
- 交替分支按书写顺序优先尝试，而非最长优先：`a|aa|aaa` 在 `aaaa` 上得到 4 个 `a`，
  `aaa|aa|a` 则得到 `aaa` 与 `a` 两个结果。
- 量词：`*` 为 0 次以上、`+` 为 1 次以上、`?` 为 0 或 1 次、`{2}` 恰为 2 次、`{1,2}` 为 1 至 2 次。
  上限决定能否匹配：`ca{1,2}t` 无法匹配含 3 个 a 的 `caaat`。

### 实验 2 分组与替换说明

- 捕获组用括号定义，`findall` 在存在两个及以上分组时返回元组列表，顺序与括号顺序一致。
- `\1`、`\2` 为后向引用，要求与对应分组**内容完全相同**，用于查找连续重复的单词；
  必须使用原始字符串，否则 `\1` 会被解释为八进制转义。
- `\2` 在日期任务中引用分隔符，保证「年-月-日」三处的分隔符一致；
  替换串 `\4-\3-\1` 负责重排为「日-月-年」，`\n` 不在匹配范围内，故行数保持不变。
- 电话号码任务通过 `re.sub` 与三组捕获完成格式重排，`\b` 用于防止在更长数字串中截取片段。

### 实验 3 解析结果说明

- `BeautifulSoup(html_doc, "html.parser")` 构建树形结构，`html.parser` 为 Python 内置解析器。
- `soup.title.text` 取得标题文本；`soup.h1.name` 为标签名 `h1`，`.text` 为节点文本。
- `find_all("p", class_="course")` 因 `class` 是 Python 关键字而使用下划线后缀；
  `get('class')` 返回列表，说明 class 属性支持多值。
- `get_text()` 递归拼接所有后代文本，不自动添加分隔符；`<title>` 属于文档文本因而被计入，
  若要页面正文应使用 `soup.body.get_text()`。

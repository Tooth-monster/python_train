作业1：字符串的乐趣

```python
#1-1 is_palindrome(s)：忽略空格和大小写，判断回文。

def is_palindrome(s):
    """忽略空格与大小写，判断 s 是否为回文。"""
    # 1) 去掉所有空格  2) 统一转小写
    cleaned = s.replace(" ", "").lower()
    # 3) 正序与反序相同即为回文
    return cleaned == cleaned[::-1]


# ---- 自测 ----
for case in PALINDROME_CASES:
    print(f"{case!r:<36} -> {is_palindrome(case)}")
```

```python
#1-2 is_abecedarian(s)：字母从左到右按字母序排列即返回 True， 允许重复，忽略空格。
def is_abecedarian(s):
    """忽略空格，判断字母是否按升序排列（允许重复）。"""
    letters = s.replace(" ", "").lower()
    # 任意相邻两个字符，前一个小于等于后一个，则整体有序
    return all(letters[i] <= letters[i + 1] for i in range(len(letters) - 1))


# ---- 自测 ----
for case in ABECEDARIAN_CASES:
    print(f"{case!r:<16} -> {is_abecedarian(case)}")
```

```python
#1-3 remove_vowels(s)：去掉所有元音 a、e、i、o、u。
VOWELS = set("aeiou")

def remove_vowels(s):
    """去除 s 中所有的 a/e/i/o/u（保留大小写原样）。"""
    return "".join(ch for ch in s if ch.lower() not in VOWELS)


# ---- 自测 ----
for case in REMOVE_VOWEL_CASES:
    print(f"{case!r:<16} -> {remove_vowels(case)!r}")

# 全元音字符串
print(f"{'aeiou'!r:<16} -> {remove_vowels('aeiou')!r}   （应为空串 ''）")
```

作业2：列表的乐趣

```python
#2-1 list_reverse(lst)：返回反转后的列表，输入不是列表要报错。
def list_reverse(lst):
    """返回 lst 的反转副本；lst 不是列表时抛 TypeError。"""
    if not isinstance(lst, list):
        raise TypeError(f"list_reverse 需要一个 list，收到的是 {type(lst).__name__}")
    return lst[::-1]          # 切片反转，返回新列表


# ---- 自测 ----
print(list_reverse([1, 2, 3]))          # [3, 2, 1]
print(list_reverse([]))                 # []
print(list_reverse(['a', 'b']))         # ['b', 'a']

# 原列表未被修改
nums = [1, 2, 3]
print("原列表：", nums, "| 反转：", list_reverse(nums), "| 再看原列表：", nums)

# 类型错误测试
for bad in [(1, 2, 3), "abc", 123, None]:
    try:
        list_reverse(bad)
    except TypeError as e:
        print(f"TypeError: {e}")
```

```python
#2-2 is_sorted(seq)：序列按升序返回 True， 只能遍历一次。
def is_sorted(seq):
    """一次遍历判断 seq 是否按升序排列（允许相等）。"""
    for i in range(len(seq) - 1):
        if seq[i] > seq[i + 1]:      # 发现一次逆序就立刻返回
            return False
    return True


# ---- 自测 ----
print(is_sorted([1, 2, 3, 4]))       # True
print(is_sorted([1, 2, 2, 3]))       # True  （允许相等）
print(is_sorted([1, 3, 2]))          # False
print(is_sorted([]))                 # True  （空序列）
print(is_sorted([5]))                # True  （单元素）

# 题目强调的动态类型：不同的序列类型都能用
print(is_sorted("abcde"))            # True
print(is_sorted((1, 2, 3)))          # True
print(is_sorted("abdc"))             # False
```

作业3：作为向量的元组

```python
#3-1 vec_scalar_mult(t, s)：元组 × 标量，返回同长度元组。 类型不合适抛 TypeError。

def _is_number(x):
    return isinstance(x, (int, float)) and not isinstance(x, bool)


def vec_scalar_mult(t, s):
    """返回元组 t 的每个元素乘以标量 s 后的新元组。"""
    if not isinstance(t, tuple):
        raise TypeError(f"第一个参数必须是 tuple，收到 {type(t).__name__}")
    if not _is_number(s):
        raise TypeError(f"第二个参数必须是 int 或 float，收到 {type(s).__name__}")
    if not all(_is_number(x) for x in t):
        raise TypeError("元组 t 的每个元素都必须是 int 或 float")
    return tuple(x * s for x in t)


# ---- 自测 ----
print(vec_scalar_mult((1, 2, 3), 2))         # (2, 4, 6)
print(vec_scalar_mult((1.5, -2.0), 3))       # (4.5, -6.0)
print(vec_scalar_mult((1, 2, 3), 0.5))       # (0.5, 1.0, 1.5)
print(vec_scalar_mult((), 5))                # ()  空元组也是合法向量
print(vec_scalar_mult((True, 2), 3) if False else "跳过 bool 测试")

for bad in [([1, 2], 2), ((1, 2), "2"), ("abc", 2), ((1, "a"), 2)]:
    try:
        vec_scalar_mult(*bad)
    except TypeError as e:
        print("TypeError:", e)
```

```python
#3-2 vec_inner_product(x, y)：返回内积，结果必须是 float。

def vec_inner_product(x, y):
    """返回两个元组向量的内积（float）。"""
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        raise TypeError(
            f"两个参数都必须是 tuple，收到 {type(x).__name__} 和 {type(y).__name__}"
        )
    if len(x) != len(y):
        raise ValueError(f"两个向量长度必须一致，收到 {len(x)} 和 {len(y)}")
    if not all(_is_number(v) for v in x) or not all(_is_number(v) for v in y):
        raise TypeError("向量的每个元素都必须是 int 或 float")
    # 逐个相乘再求和，最后转 float（题面要求返回浮点数）
    return float(sum(a * b for a, b in zip(x, y)))


# ---- 自测 ----
print(vec_inner_product((1, 2, 3), (4, 5, 6)))    # 4+10+18 = 32.0
print(vec_inner_product((1.5, 2), (2, 4)))        # 3.0+8 = 11.0
print(vec_inner_product((), ()))                  # 0.0

# 类型/长度错误
try:
    vec_inner_product([1, 2], (1, 2))
except TypeError as e:
    print("TypeError:", e)

try:
    vec_inner_product((1, 2), (1, 2, 3))
except ValueError as e:
    print("ValueError:", e)
```

```python
#3-3 check_valid_mx(mx)：判断是否为合法矩阵，返回布尔值 （不抛异常）

def check_valid_mx(mx):
    """合法矩阵返回 True，否则返回 False（不抛异常）。"""
    # 条件 0：本身必须是元组
    if not isinstance(mx, tuple):
        return False
    # 边界：空矩阵视为合法（0 行），这是约定，需在注释里说明
    if len(mx) == 0:
        return True

    # 条件 1：每个元素都是元组
    if not all(isinstance(row, tuple) for row in mx):
        return False

    # 条件 2：所有内层长度相同（以第一行为基准）
    n = len(mx[0])
    if any(len(row) != n for row in mx):
        return False

    # 条件 3：每个分量都是数字
    if not all(_is_number(v) for row in mx for v in row):
        return False

    return True
# ---- 自测 ----
valid_mx = ((1, 2, 3), (4, 5, 6))          # 2x3
print(check_valid_mx(valid_mx))            # True
print(check_valid_mx(((1, 2), (3, 4))))    # True
print(check_valid_mx(((1.5, 2), (3, 4))))  # True
print(check_valid_mx(()))                  # True  空矩阵

print(check_valid_mx([[1, 2], [3, 4]]))    # False 外层是 list 不是 tuple
print(check_valid_mx(((1, 2), (3, 4, 5))))  # False 行长不一致
print(check_valid_mx(((1, 2), (3, "a"))))  # False 含非数字
print(check_valid_mx((1, 2, 3)))           # False 外层元素不是 tuple
print(check_valid_mx("abc"))               # False 根本不是元组
```

```python
#3-4 mx_vec_mult(mx, v)：矩阵乘向量，返回数值元组。 用 check_valid_mx 校验，不合法要抛错。
def mx_vec_mult(mx, v):
    """m×n 矩阵（元组的元组）乘 n 维向量（元组），返回 m 维元组。"""
    # ---- 参数校验 ----
    if not check_valid_mx(mx):
        raise ValueError("第一个参数不是合法矩阵，请用 check_valid_mx 检查")
    if not isinstance(v, tuple):
        raise TypeError(f"第二个参数必须是 tuple，收到 {type(v).__name__}")
    if not all(_is_number(x) for x in v):
        raise TypeError("向量 v 的每个元素都必须是 int 或 float")
    if len(mx) == 0:
        raise ValueError("矩阵不能为空（0 行无法确定列数）")

    n_cols = len(mx[0])
    if len(v) != n_cols:
        raise ValueError(
            f"矩阵列数({n_cols})与向量维度({len(v)})不一致，无法相乘"
        )

    # ---- 逐行与 v 做内积 ----
    return tuple(
        float(sum(row[j] * v[j] for j in range(n_cols)))
        for row in mx
    )
# ---- 自测 ----
M = ((1, 2, 3),
     (4, 5, 6))            # 2x3
u = (1, 1, 1)              # 3 维
print("M @ u =", mx_vec_mult(M, u))     # (6.0, 15.0)

u2 = (2, 0, -1)
print("M @ u2 =", mx_vec_mult(M, u2))   # (1*2+2*0+3*(-1), 4*2+5*0+6*(-1)) = (-1.0, 2.0)

# 单位矩阵验证：I @ v == v
I = ((1, 0), (0, 1))
print("I @ (7, 9) =", mx_vec_mult(I, (7, 9)))   # (7.0, 9.0)

# ---- 错误测试 ----
for bad_mx, bad_v in [(((1, 2), (3, 4, 5)), (1, 2)),       # 矩阵本身不合法
                      (((1, 2), (3, 4)), (1, 2, 3)),       # 列数与向量维度不符
                      (((1, 2), (3, 4)), [1, 2])]:         # 向量不是元组
    try:
        mx_vec_mult(bad_mx, bad_v)
    except (TypeError, ValueError) as e:
        print(f"{type(e).__name__}: {e}")

```

作业4：更多向量的乐趣（面向对象）

```python
#4-1 定义 Vector 类：维度 + 元素，带完整校验和默认值。

class Vector:
    """用维度 + 元素表示的向量。"""

    def __init__(self, dim, elements=None):
        # ---- 校验维度 ----
        if not isinstance(dim, int) or isinstance(dim, bool):
            raise TypeError(f"维度必须是 int，收到 {type(dim).__name__}")
        if dim < 0:
            raise ValueError(f"维度必须是非负整数，收到 {dim}")

        # ---- 只给维度：默认零向量 ----
        if elements is None:
            elements = [0] * dim

        # ---- 校验元素容器类型 ----
        if not isinstance(elements, (list, tuple)):
            raise TypeError(
                f"元素必须是 list 或 tuple，收到 {type(elements).__name__}"
            )
        # ---- 校验长度一致 ----
        if len(elements) != dim:
            raise ValueError(
                f"维度({dim})与元素个数({len(elements)})不一致"
            )
        # ---- 校验每个元素是数字 ----
        if not all(_is_number(x) for x in elements):
            raise TypeError("向量的每个元素都必须是 int 或 float")

        # 统一存成元组，防止外部通过列表引用改掉内部状态
        self.dim = dim
        self.elements = tuple(float(x) for x in elements)

    # 让 print 出来好看（调试必备，建议先写）
    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"
# ---- 自测 ----
v1 = Vector(3, (1, 2, 3))
v2 = Vector(3, [4, 5, 6])
v3 = Vector(4)                    # 只给维度 → 零向量 (0,0,0,0)
v4 = Vector(0)                    # 0 维向量
print(v1, v2, v3, v4)

# 各类错误
for args in [(3.0, (1, 2, 3)),        # 维度不是 int
             (-1, []),                # 维度为负
             (True, [1]),             # bool 不是合法维度
             (3, (1, 2)),             # 长度不一致
             (2, "ab"),               # 元素不是 list/tuple
             (2, (1, "a"))]:          # 元素含非数字
    try:
        Vector(*args)
    except (TypeError, ValueError) as e:
        print(f"{type(e).__name__}: {e}")
    
```

4-2 维度与元素应该是类属性还是实例属性？说明理由。

它们是每个对象各自的数据。类属性被所有实例共享， 写 Vector.dim = 3 会让所有向量都变成 3 维——这与"每个向量有自己的维度" 直接矛盾。Vector(2, (1, 2)) 和 Vector(3, (1, 2, 3)) 必须能同时存在。

修改要互不影响。若 elements 定义在类上，所有实例指向 同一个可变对象，改一个向量的分量会连带改掉全部（Python 著名的 "可变类属性"陷阱）。

类属性的唯一合理用途是常量。比如把 _EPSILON = 1e-9 或允许的元素类型放进类属性，因为所有实例共用一个值 且不需要修改；而 dim、elements 属于"随对象而变"的状态。

```python
#4-3 实现比较运算符：==、<、<=、 >（含 >= 和 !=）。按元组字典序比较， 维度不同抛异常
from functools import total_ordering

@total_ordering                     # 只需实现 __eq__ 和 __lt__，其余自动生成
class Vector:
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int) or isinstance(dim, bool):
            raise TypeError(f"维度必须是 int，收到 {type(dim).__name__}")
        if dim < 0:
            raise ValueError(f"维度必须是非负整数，收到 {dim}")
        if elements is None:
            elements = [0] * dim
        if not isinstance(elements, (list, tuple)):
            raise TypeError(f"元素必须是 list 或 tuple，收到 {type(elements).__name__}")
        if len(elements) != dim:
            raise ValueError(f"维度({dim})与元素个数({len(elements)})不一致")
        if not all(_is_number(x) for x in elements):
            raise TypeError("向量的每个元素都必须是 int 或 float")
        self.dim = dim
        self.elements = tuple(float(x) for x in elements)

    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"

    # ---- 内部辅助：比较前先校验维度一致 ----
    def _check_comparable(self, other):
        if not isinstance(other, Vector):
            raise TypeError(f"只能与 Vector 比较，收到 {type(other).__name__}")
        if self.dim != other.dim:
            raise ValueError(
                f"维度不同的向量不能比较：{self.dim} vs {other.dim}"
            )

    # ---- 相等 ----
    def __eq__(self, other):
        self._check_comparable(other)
        return self.elements == other.elements

    # ---- 小于：直接用元组的字典序 ----
    def __lt__(self, other):
        self._check_comparable(other)
        return self.elements < other.elements


# ---- 自测 ----
print("(2,4) <  (2,5) :", Vector(2, (2, 4)) <  Vector(2, (2, 5)))   # True
print("(2,4) == (2,4) :", Vector(2, (2, 4)) == Vector(2, (2, 4)))   # True
print("(2,4) <= (2,4) :", Vector(2, (2, 4)) <= Vector(2, (2, 4)))  # True
print("(2,5) >  (2,4) :", Vector(2, (2, 5)) >  Vector(2, (2, 4)))   # True
print("(2,4) != (9,9) :", Vector(2, (2, 4)) != Vector(2, (9, 9)))   # True

print("排序：", sorted([Vector(2, (3, 1)), Vector(2, (1, 9)), Vector(2, (3, 0))]))

# 维度不同要抛异常
try:
    Vector(2, (1, 2)) < Vector(3, (1, 2, 3))
except ValueError as e:
    print("ValueError:", e)


```

```python
class Vector:
    def __init__(self, dim, data):
        self.dim = dim
        self.elements = list(data)   # 改成 elements，和dot里保持一致

    #4-4 实现 Vector.dot(self, other)：返回内积，类型或维度不对 要抛异常。
    def dot(self, other):
        """返回 self 与 other 的内积（float）。"""
        if not isinstance(other, Vector):
            raise TypeError(f"dot 的参数必须是 Vector，收到 {type(other).__name__}")
        if self.dim != other.dim:
            raise ValueError(f"维度不一致，无法求内积：{self.dim} vs {other.dim}")
        return float(sum(a * b for a, b in zip(self.elements, other.elements)))

# ---- 自测 ----
a = Vector(3, (1, 2, 3))
b = Vector(3, (4, 5, 6))
print("a.dot(b) =", a.dot(b))                # 4+10+18 = 32.0
print("类型：", type(a.dot(b)).__name__)      # float
try:
    a.dot((4, 5, 6))                         # 传元组不是 Vector
except TypeError as e:
    print("TypeError:", e)
try:
    a.dot(Vector(2, (1, 2)))                 # 维度不一致
except ValueError as e:
    print("ValueError:", e)
```

```python
class Vector:
    def __init__(self, dim, data):
        self.dim = dim
        self.elements = list(data)

    def dot(self, other):
        """返回 self 与 other 的内积（float）。"""
        if not isinstance(other, Vector):
            raise TypeError(f"dot 的参数必须是 Vector，收到 {type(other).__name__}")
        if self.dim != other.dim:
            raise ValueError(f"维度不一致，无法求内积：{self.dim} vs {other.dim}")
        return float(sum(a * b for a, b in zip(self.elements, other.elements)))

    def __mul__(self, other):
        """
        处理 v * other
        情况1：other是标量 → 标量右乘，返回新Vector
        情况2：other是Vector → 逐元素相乘（哈达玛积），维度不同抛异常
        """
        if isinstance(other, (int, float)):
            new_elems = [x * other for x in self.elements]
            return Vector(self.dim, new_elems)
        elif isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError(f"逐元素相乘维度不一致：{self.dim} vs {other.dim}")
            new_elems = [a * b for a, b in zip(self.elements, other.elements)]
            return Vector(self.dim, new_elems)
        else:
            raise TypeError(f"* 不支持 Vector 和 {type(other).__name__} 相乘")

    def __rmul__(self, other):
        """处理 other * v，实现标量左乘，例如 2 * v"""
        # 标量左乘和右乘逻辑一样，直接复用 __mul__
        return self * other

    def __repr__(self):
        return f"Vector({self.dim}, {tuple(self.elements)})"


# ========== 自测代码 ==========
if __name__ == "__main__":
    # 标量右乘 v*2
    v1 = Vector(2, (2, 3))
    print("v1 * 2 =", v1 * 2)

    # 标量左乘 2*v
    print("2 * v1 =", 2 * v1)

    # 向量逐元素相乘 v * w
    v2 = Vector(2, (4, 5))
    print("v1 * v2 =", v1 * v2)  # (8,15)

    # 维度不匹配测试
    try:
        v1 * Vector(3, (1, 2, 3))
    except ValueError as e:
        print("ValueError:", e)

    # 原来dot方法自测
    a = Vector(3, (1, 2, 3))
    b = Vector(3, (4, 5, 6))
    print("a.dot(b) =", a.dot(b))
    print("类型：", type(a.dot(b)).__name__)
    try:
        a.dot((4, 5, 6))
    except TypeError as e:
        print("TypeError:", e)
    try:
        a.dot(Vector(2, (1, 2)))
    except ValueError as e:
        print("ValueError:", e)
```
